```python
from astropy.cosmology import FlatLambdaCDM
import numpy as np
import os

if os.path.exists("sep.txt"):
    os.remove("sep.txt")
    os.system("touch sep.txt")
if os.path.exists("ngVLA_res_1.txt"):
    os.remove("ngVLA_res_1.txt")
    os.system("touch ngVLA_res_1.txt")
if os.path.exists("ngVLA_res_2.txt"):
    os.remove("ngVLA_res_2.txt")
    os.system("touch ngVLA_res_2.txt")
if os.path.exists("vla_max.txt"):
    os.remove("vla_max.txt")
    os.system("touch vla_max.txt")

# Cosmology
H0 = 70
Om0 = 0.29
cosmo = FlatLambdaCDM(H0=H0, Om0=Om0)

# Redshifts
z_vals = np.logspace(np.log10(0.001), np.log10(10), 1000)

# Angular scale: 1 mas in radians
mas = 1e-3 / 3600 * np.pi / 180  # 1 mas -> rad
ngVLA_1 = 0.08*1e-3  # ngVLA 93 GHz resolution in arcsec
ngVLA_2 = 2.97*1e-3  # ngVLA 2.4 GHz resolution in arcsec
vla_max = 0.0645  # VLA max resolution in arcsec

for z in z_vals:
    # kpc per arcsec
    kpc_per_arcsec = cosmo.kpc_proper_per_arcmin(z).value / 60
    # linear scale for 1 mas
    linear_pc = kpc_per_arcsec # multiply by 1000 for kpc to pc and divide by 1000 for mas to arcsec
    print(f"{z:.5f}\t{linear_pc:.5f}")
    
    # linear scale for ngVLA resolutions
    linear_pc_1 = linear_pc * ngVLA_1 * 1e3
    linear_pc_2 = linear_pc * ngVLA_2 * 1e3
    linear_pc_vla_max = linear_pc * vla_max * 1e3
    print(f"{z:.5f}\t{linear_pc_1:.5f}\t{linear_pc_2:.5f}")

    # Write results to files
    os.system(f"echo '{z:.5f}\t{linear_pc:.5f}' >> sep.txt")
    os.system(f"echo '{z:.5f}\t{linear_pc_1:.5f}' >> ngVLA_res_1.txt")
    os.system(f"echo '{z:.5f}\t{linear_pc_2:.5f}' >> ngVLA_res_2.txt")
    os.system(f"echo '{z:.5f}\t{linear_pc_vla_max:.5f}' >> vla_max.txt")
```
